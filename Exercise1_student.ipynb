## <font color='#05c3de'>loc - iloc</font>  

- <code>loc</code>: selects data by **labels**
- <code>iloc</code>: selects data by **integer positions**  

- The numbers displayed on the left are index labels; with the default index, labels and positions happen to look the same.

## <code>loc</code>   
### data.loc [row_labels, column_labels]

🧩 Let's select the value in the row with label 0 and the Rented_bike_count column.

```python
df.loc[0,"Rented_bike_count"]
```

🧩 Let's select the Date, Rented_bike_count, and Temperature columns for rows with labels 0 to 5.

```python
df.loc[:5, ["Date", "Rented_bike_count", "Temperature"]]
```

💡 If the slice starts from the beginning, the first label can be omitted.   
💡 With <code>loc</code>, the end label is included.

## <code>iloc</code>
### data.iloc [row_positions, column_positions]

🧩 Let's select the value in the first row and the second column.

```python
df.iloc[0, 1]
```

🧩 Let's select the first 5 rows and the first 4 columns.

```python
df.iloc[:5, :4]
```

💡 With <code>iloc</code>, the end position is excluded.

### Quick CheatSheet

| Action                  | `loc` – By Label                        | `iloc` – By Position    |
| ----------------------- | --------------------------------------- | ----------------------- |
| Select a single row     | `data.loc[2]`                           | `data.iloc[2]`          |
| Select a single column  | `data.loc[:, "Temperature"]`            | `data.iloc[:, 3]`       |
| Select multiple rows    | `data.loc[[1, 3, 5]]`                   | `data.iloc[[1, 3, 5]]`  |
| Select multiple columns | `data.loc[:, ["Date", "Temperature"]]`  | `data.iloc[:, [0, 3]]`  |
| Select a range of rows  | `data.loc[:5]`                          | `data.iloc[:5]`         |
| Select rows and columns | `data.loc[:5, ["Date", "Temperature"]]` | `data.iloc[:5, [0, 3]]` |
| Select from the end     | —                                       | `data.iloc[-5:, -2:]`   |


<div style="background-color: #05c3de; padding: 10px;">
    <h2><font color='white'> Boolean filtering🔽</font></h2>
</div>

- #### We can filter a DataFrame to select only rows that satisfy a specific condition.  

 A comparison asks a yes-or-no question about every row:
 
 > Is the temperature above 25 °C?

The result contains one Boolean value for every row:

- `True` — the condition is satisfied,
- `False` — the condition is not satisfied.


🧩 Let's see which rows have temperatures above 25 °C.

```python
is_hot = df["Temperature"] > 25
is_hot.head()
```

💡 The condition returns a Boolean `Series` containing `True` or `False` for each row of `df`.

```python
is_hot.value_counts()
```

`value_counts()` shows how many rows satisfy the condition and how many do not.

### Applying the Boolean mask

When a Boolean Series is used to select rows:

- rows with `True` are kept,
- rows with `False` are excluded from the result.

```python
df[is_hot]
```

With loc, we can filter rows and select columns at the same time.

```python
df.loc[is_hot,["Date", "Hour", "Temperature", "Rented_bike_count"]].head()
```

💡 Pandas checks the condition for each value in the selected column and returns True or False.   
When the mask is applied to the DataFrame, only the rows where the mask is True are kept.

### Cheat Sheet: Comparison Operators  

| Operator | Meaning                  |

| `!=`     | Not equal to             |
| `>`      | Greater than             |
| `<`      | Less than                |
| `>=`     | Greater than or equal to |
| `<=`     | Less than or equal to    |

#### Combining Multiple Conditions  

| Operator | Meaning |
| -------- | ------- |
| `&`      | AND     |
| `\|`     | OR      |
| `~`      | NOT     |

🧩 Look at the following code.  

- How would you describe following condition?
- When will `is_warm_and_dry` contain `True`?
- Does the code already filter the DataFrame?

```python
is_warm = df["Temperature"] > 20
is_dry = df["Rainfall"] == 0

is_warm_and_dry = is_warm & is_dry
```

The `&` operator means AND. The resulting mask is True for a row only when both conditions are True for that row.

💡It only creates the Boolean mask. It does not filter the DataFrame yet.

```python
df.loc[is_warm_and_dry,["Date", "Hour", "Temperature", "Rainfall"]].head()
```

more compact code:

```python
is_warm_and_dry = ((df["Temperature"] > 20) & (df["Rainfall"] == 0))
```

💡 When combining multiple conditions, put each condition in parentheses ().

💡 With loc, we can use a Boolean mask to filter rows and select specific columns at the same time.
